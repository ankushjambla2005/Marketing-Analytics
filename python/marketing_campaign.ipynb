```python
import pandas as pd
df_nykaa=pd.read_csv(r"C:\project placement\csv dataset raw\nykaa_campaign_data.csv")
df_purplle=pd.read_csv(r"C:\project placement\csv dataset raw\purplle_campaign_data.csv")
df_tira=pd.read_csv(r"C:\project placement\csv dataset raw\tira_campaign_data.csv")
```

```python
df_nykaa.info()
```

```python
#Here watching the data we see that Date is in str now we will change it to datetime 
df_nykaa["Date"]=pd.to_datetime(df_nykaa.Date,format='mixed')
df_purplle["Date"]=pd.to_datetime(df_purplle.Date,format='mixed')
df_tira["Date"]=pd.to_datetime(df_tira.Date,format='mixed')
df_purplle.info()
df_tira.info()
```

```python
#will change every column heading to lower case
df_nykaa.columns=df_nykaa.columns.str.lower()
df_purplle.columns=df_purplle.columns.str.lower()
df_tira.columns=df_tira.columns.str.lower()
```

```python
df_nykaa.channel_used.value_counts()
```

```python
#combine the whole three table nykaa,purplle,tira
df_master=pd.concat([df_nykaa,df_purplle,df_tira],ignore_index=True)
```
