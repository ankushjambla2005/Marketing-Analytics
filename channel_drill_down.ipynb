```python
import pandas as pd
df_nykaa=pd.read_csv(r"C:\project placement\csv dataset raw\nykaa_campaign_data.csv")
df_purplle=pd.read_csv(r"C:\project placement\csv dataset raw\purplle_campaign_data.csv")
df_tira=pd.read_csv(r"C:\project placement\csv dataset raw\tira_campaign_data.csv")
```

```python
df_nykaa.columns=df_nykaa.columns.str.lower()
df_purplle.columns=df_nykaa.columns.str.lower()
df_tira.columns=df_nykaa.columns.str.lower()
```

```python
df_nykaa["date"]=pd.to_datetime(df_nykaa.date ,format='mixed')
df_purplle["date"]=pd.to_datetime(df_purplle.date ,format='mixed')
df_tira["date"]=pd.to_datetime(df_tira.date ,format='mixed')
```

```python
#we can see the channel_used contains multiple value in single row
df_nykaa
```

```python
df_nykaa["channel_used"]=df_nykaa.channel_used.str.split(',')
df_nykaa_exploded=df_nykaa.explode("channel_used")
```

```python
df_nykaa_exploded[["campaign_id","channel_used"]]
```

```python
df_purplle["channel_used"]=df_purplle.channel_used.str.split(',')
df_purplle_exploded=df_purplle.explode("channel_used")
df_tira["channel_used"]=df_tira.channel_used.str.split(',')
df_tira_exploded=df_tira.explode("channel_used")
```

```python
df_purplle_exploded[["campaign_id","channel_used"]]
```

```python
df_nykaa_exploded["channel_used"]=df_nykaa_exploded.channel_used.str.strip()
df_purplle_exploded["channel_used"]=df_purplle_exploded.channel_used.str.strip()
df_tira_exploded["channel_used"]=df_tira_exploded.channel_used.str.strip()
```

```python
df_nykaa_exploded=df_nykaa_exploded[["campaign_id","channel_used","brand"]]
```

```python
df_purplle_exploded=df_purplle_exploded[["campaign_id","channel_used","brand"]]
df_tira_exploded=df_tira_exploded[["campaign_id","channel_used","brand"]]
```

```python
df_master_exploded=pd.concat([df_nykaa_exploded,df_purplle_exploded,df_tira_exploded],ignore_index=True)
```
